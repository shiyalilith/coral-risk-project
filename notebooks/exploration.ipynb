# Coral Risk Exploration

Use this notebook for quick inspection and plots. Training logic lives in `src/`.

```python
from pathlib import Path

import matplotlib.pyplot as plt

from src.data_loader import load_sst_frame
from src.features import build_model_frame
```

```python
data_path = Path("data/raw/noaaSSTcoralData.nc")
raw_frame = load_sst_frame(data_path)
model_frame = build_model_frame(raw_frame, forecast_horizon=14, risk_delta=0.75)

raw_frame.head()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

axes[0].plot(model_frame["time"], model_frame["sst"], label="SST")
axes[0].plot(model_frame["time"], model_frame["sst_baseline"], label="Baseline")
axes[0].set_ylabel("Temperature (C)")
axes[0].legend()

axes[1].plot(model_frame["time"], model_frame["rolling_heat_stress"], label="Rolling heat stress", color="tab:red")
axes[1].scatter(
    model_frame.loc[model_frame["future_risk"].eq(1), "time"],
    model_frame.loc[model_frame["future_risk"].eq(1), "rolling_heat_stress"],
    color="black",
    s=12,
    label="Future risk",
)
axes[1].set_ylabel("Stress")
axes[1].legend()
plt.tight_layout()
```
